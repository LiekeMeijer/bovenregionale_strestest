```python
from pathlib import Path
import numpy as np
import pandas as pd
import geopandas as gpd
```

```python
region_list = [
    "Groningen en NO-Drenthe", "Noord-Westelijke Delta", "Overijsselse Vecht",
    "Limburg", "Vallei en Veluwe", "Achterhoek", "Brabantse Delta",
    "Friesland", "ARK-NZK", "Noord-Brabant Oost", "Rivierenland",
    "Scheldestromen", "Zuiderzeeland"
]


for region in region_list:
    root_dir = Path(rf"P:\bovenregionale-stresstest-hwn\Analysis\{region}\Outputs")
    gpkg_path = root_dir / "Aggregated_schakels.gpkg"
    roads_ex = gpd.read_file(gpkg_path)

    # Add damage per meter (safe numeric conversion + zero / NaN guard)
    for col in ["total_damage", "total_length"]:
        if col in roads_ex.columns:
            roads_ex[col] = pd.to_numeric(roads_ex[col], errors="coerce")
        else:
            raise KeyError(f"Required column '{col}' missing in {gpkg_path}")

    roads_ex["dam_per_m"] = (
        (roads_ex["total_damage"] / roads_ex["total_length"])
        .where(roads_ex["total_length"].notna() & (roads_ex["total_length"] != 0))
    )

    # Rank damage per meter (highest dam_per_m gets rank 1)
    roads_ex["rk_d_m"] = (
        roads_ex["dam_per_m"]
        .rank(ascending=False, method="min")
        .astype("Int64")
    )

    # Save ranked features as GeoPackage
    out_gpkg = root_dir / "ranked_Aggregated_schakels.gpkg"
    try:
        roads_ex.to_file(out_gpkg, layer="ranked_rows", driver="GPKG", if_exists="replace")
    except TypeError:
        if out_gpkg.exists():
            out_gpkg.unlink()
        roads_ex.to_file(out_gpkg, layer="ranked_rows", driver="GPKG")
    print(f"Wrote: {out_gpkg}")

```

```python
#Rank for Losses

from pathlib import Path
import warnings
import re
import pandas as pd
import geopandas as gpd

area_gpkg = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Areas.gpkg")
gdf_area = gpd.read_file(area_gpkg)
input_path = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Losses_Analysis\Simple_Losses_Analysis_Schakel_level_Lieke_filtered.gpkg")
gdf_src = gpd.read_file(input_path)

def add_area_overlay(gdf: gpd.GeoDataFrame, area_path: Path, target_crs: str, key_col: str) -> gpd.GeoDataFrame:
    import fiona
    if not area_path.exists():
        warnings.warn(f"Area file not found: {area_path}. Skipping area overlay.")
        return gdf.copy()
    layers = fiona.listlayers(str(area_path))
    layer = layers[0]
    area = gpd.read_file(area_path, layer=layer)
    if area.crs is None:
        area.set_crs(target_crs, inplace=True)
    if area.crs.to_string() != target_crs:
        area = area.to_crs(target_crs)
    candidate_labels = ["name", "id"]
    label_col = next((c for c in candidate_labels if c in area.columns), None)
    new_col = area_path.stem  # "Areas"
    left = gdf[[key_col, "geometry"]].copy()
    right_cols = ["geometry"] + ([label_col] if label_col else [])
    sj = gpd.sjoin(left, area[right_cols], how="left", predicate="intersects")
    has_overlap = (
        sj.groupby(key_col)["index_right"]
          .apply(lambda s: s.notna().any())
          .rename(new_col)
          .reset_index()
    )
    out = gdf.merge(has_overlap, on=key_col, how="left")
    out[new_col] = out[new_col].fillna(False)
    if label_col:
        labels = (
            sj.groupby(key_col)[label_col]
              .apply(lambda s: ";".join(sorted({str(v) for v in s.dropna()})) if s.notna().any() else None)
              .rename(f"{new_col}_{label_col}")  # "Areas_name" if label_col == "name"
              .reset_index()
        )
        out = out.merge(labels, on=key_col, how="left")
    return out

def split_areas(val):
    if pd.isna(val):
        return []
    parts = [p.strip() for p in re.split(r"[;,\u061B]", str(val))]
    return [p for p in parts if p]

TARGET_CRS = "EPSG:28992"
area_gpkg = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Areas.gpkg")
merged = add_area_overlay(gdf_src, area_gpkg, TARGET_CRS, key_col="NETWERKSCH_HWN")

# Ensure required columns
if "Areas_name" not in merged.columns:
    raise KeyError("Column 'Areas_name' not found after overlay. Make sure Areas.gpkg has a 'name' field.")

if "VHLH_AL_E_WR" not in merged.columns:
    raise KeyError("Column 'VHLH_AL_E_WR' not found in losses input.")

# Prepare, explode per Area, and rank
merged["VHLH_AL_E_WR"] = pd.to_numeric(merged["VHLH_AL_E_WR"], errors="coerce")

gdf_loss_exp = (
    merged.assign(Area=merged["Areas_name"].apply(split_areas))
          .explode("Area", ignore_index=True)
)

gdf_loss_exp = gdf_loss_exp[gdf_loss_exp["Area"].notna() & (gdf_loss_exp["Area"] != "")]

gdf_loss_exp["rk_VHLH_AL_E_WR"] = (
    gdf_loss_exp.groupby("Area")["VHLH_AL_E_WR"].rank(ascending=False, method="min")
)

# Keep all columns; sort for readability
gdf_loss_ranked = (
    gdf_loss_exp
      .sort_values(["Area", "rk_VHLH_AL_E_WR", "VHLH_AL_E_WR"], ascending=[True, True, False], kind="mergesort")
      .reset_index(drop=True)
)

# Write outputs
base_dir = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs")
base_dir.mkdir(parents=True, exist_ok=True)
gpkg_path = base_dir / "Simple_Losses_ranking.gpkg"
shp_path  = base_dir / "Simple_Losses_ranking.shp"

# GPKG (keep all columns, full names)
try:
    gdf_loss_ranked.to_file(gpkg_path, layer="ranked_rows", driver="GPKG", if_exists="replace")
except TypeError:
    if gpkg_path.exists():
        gpkg_path.unlink()
    gdf_loss_ranked.to_file(gpkg_path, layer="ranked_rows", driver="GPKG")

# SHP: cast rank to int and shorten the rank field name to <=10 chars
gdf_loss_ranked_shp = gdf_loss_ranked.copy()
gdf_loss_ranked_shp["rk_VHLH_AL_E_WR"] = gdf_loss_ranked_shp["rk_VHLH_AL_E_WR"].fillna(-1).astype("int32")
rank_shp_name = "rk_VHLH"  # shapefile field name limit
if rank_shp_name != "rk_VHLH_AL_E_WR":
    gdf_loss_ranked_shp = gdf_loss_ranked_shp.rename(columns={"rk_VHLH_AL_E_WR": rank_shp_name})

# Clean old shapefile sidecars
def delete_shapefile(path: Path):
    for ext in [".shp", ".shx", ".dbf", ".prj", ".cpg", ".shp.xml", ".qpj"]:
        p = path.with_suffix(ext)
        if p.exists():
            p.unlink()

delete_shapefile(shp_path)
gdf_loss_ranked_shp.to_file(shp_path, driver="ESRI Shapefile")

print(f"Wrote:\n- {gpkg_path}\n- {shp_path}")
```

```python
#Rank for damages

from pathlib import Path
import pandas as pd
import geopandas as gpd
import re

input_path = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs\Aggregated_schakels_merged_with_area_edited_ver03.gpkg")
gdf_src = gpd.read_file(input_path)

# 02) Explode Areas_name (split on ';' and also allow commas just in case)
def split_areas(val):
    if pd.isna(val):
        return []
    parts = [p.strip() for p in re.split(r"[;,\u061B]", str(val))]  # ; , and Arabic semicolon
    return [p for p in parts if p]

if "Areas_name" not in gdf_src.columns:
    raise KeyError("Column 'Areas_name' not found.")

gdf_exp = (
    gdf_src.assign(Area=gdf_src["Areas_name"].apply(split_areas))
           .explode("Area", ignore_index=True)
)
gdf_exp = gdf_exp[gdf_exp["Area"].notna() & (gdf_exp["Area"] != "")]

# Group by Area (example aggregations)
by_area = (
    gdf_exp
    .groupby("Area", as_index=False)
)

by_area.head()

gdf_src["D_per_m"] = pd.to_numeric(gdf_src["D_per_m"], errors="coerce")
gdf_src["fraction_flooded"] = pd.to_numeric(gdf_src["fraction_flooded"], errors="coerce")
gdf_exp["rank_d"] = gdf_exp.groupby("Area")["D_per_m"].rank(ascending=False, method="min")
gdf_exp["rank_frfl"] = gdf_exp.groupby("Area")["fraction_flooded"].rank(ascending=False, method="min")

# Write ranked original rows (one feature per Area entry)
from pathlib import Path

required = ["Area", "rank_d", "rank_frfl", "type", "D_per_m", "NETWERKSCH_HWN", "total_damage", "fraction_flooded","total_length","flooded_length", "geometry"]
missing = [c for c in required if c not in gdf_exp.columns]
if missing:
    raise KeyError(f"Missing in gdf_exp: {missing}")

gdf_ranked = (
    gdf_exp.loc[:, required]
            .sort_values(["Area", "rank_d", "rank_frfl", "D_per_m", "fraction_flooded"],
                         ascending=[True, True, True, False, False], kind="mergesort")
            .reset_index(drop=True)
)

base_dir = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs")
base_dir.mkdir(parents=True, exist_ok=True)
gpkg_path = base_dir / "Damage_risk_ranking_ver03.gpkg"
shp_path  = base_dir / "Damage_risk_ranking_ver03.shp"

# GPKG (full field names)
gdf_ranked.to_file(gpkg_path, layer="ranked_rows", driver="GPKG")

# SHP (shorten field names to <=10 chars; cast ranks to int for SHP)
gdf_ranked_shp = gdf_ranked.rename(columns={
    "NETWERKSCH_HWN": "NETW_HWN",
    "total_damage": "tot_dam",
    "fraction_flooded": "frac_fld",
}).copy()
gdf_ranked_shp["rank_d"] = gdf_ranked_shp["rank_d"].astype("Int64")
gdf_ranked_shp["rank_frfl"] = gdf_ranked_shp["rank_frfl"].astype("Int64")

gdf_ranked_shp.to_file(shp_path, driver="ESRI Shapefile")

print(f"Wrote:\n- {gpkg_path}\n- {shp_path}")
```

```python
damages = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs\Damage_risk_ranking_ver03.gpkg")
losses = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs\Simple_Losses_ranking.gpkg")

from pathlib import Path
import geopandas as gpd
import pandas as pd

# Read inputs (uses the paths defined earlier)
gdf_dam = gpd.read_file(damages, layer="ranked_rows")  # Damage_risk_ranking_ver03.gpkg
gdf_loss = gpd.read_file(losses, layer="ranked_rows")  # Simple_Losses_ranking.gpkg

# Normalize join keys
for df in (gdf_dam, gdf_loss):
    df["Area"] = df["Area"].astype(str).str.strip()
    df["NETWERKSCH_HWN"] = df["NETWERKSCH_HWN"].astype(str).str.strip()

# Keep geometry from damages; drop geometry from losses to avoid duplicate geometry columns
right_no_geom = gdf_loss.drop(columns=[c for c in gdf_loss.columns if c == "geometry"])

joined = gdf_dam.merge(
    right_no_geom,
    on=["Area", "NETWERKSCH_HWN"],
    how="inner",
    suffixes=("_dmg", "_loss"),
)

gdf_joined = gpd.GeoDataFrame(joined, geometry="geometry", crs=gdf_dam.crs)

# Write output GeoPackage
out_dir = Path(r"P:\bovenregionale-stresstest-hwn\Analysis\Merged_Outputs")
out_dir.mkdir(parents=True, exist_ok=True)
out_gpkg = out_dir / "Damages_Losses_joined_ranking.gpkg"

try:
    gdf_joined.to_file(out_gpkg, layer="joined", driver="GPKG", if_exists="replace")
except TypeError:
    if out_gpkg.exists():
        out_gpkg.unlink()
    gdf_joined.to_file(out_gpkg, layer="joined", driver="GPKG")

print(f"Joined rows: {len(gdf_joined)}")
print(f"Wrote:\n- {out_gpkg}")
```
